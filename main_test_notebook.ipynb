```python
#### python
# filepath: c:\Users\Rohith.MR\PerceptaAI\main_test_notebook.ipynb
# ...existing code...
import cv2

def capture_image(filename='captured_image.jpg'):
    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        print("Could not open the camera.")
        return
    
    ret, frame = cap.read()
    if ret:
        cv2.imwrite(filename, frame)
        print(f"Image saved as {filename}")
    else:
        print("Failed to capture image.")
    
    cap.release()

# Test the function
capture_image("test_snapshot.jpg")
# ...existing code...
```
