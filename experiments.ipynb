```python
# Fast TTS using pyttsx3
import pyttsx3

def init_fast_tts():
    try:
        engine = pyttsx3.init()
        # Optimize for speed
        rate = engine.getProperty('rate')
        engine.setProperty('rate', rate + 50)  # Faster speech
        volume = engine.getProperty('volume') 
        engine.setProperty('volume', min(volume + 0.1, 1.0))  # Slightly louder
        print("✅ Fast pyttsx3 initialized")
        return engine
    except Exception as e:
        print(f"⚠️ pyttsx3 initialization failed: {e}")
        return None

# Initialize the TTS engine
tts_engine = init_fast_tts()

# Test the TTS
if tts_engine:
    test_text = "TTS system initialized and ready"
    tts_engine.say(test_text)
    tts_engine.runAndWait()
else:
    print("Failed to initialize TTS engine")
```
